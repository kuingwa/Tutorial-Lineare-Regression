# Fehlerhaftes Notebook

Dieses Notebook enthält absichtlich drei Fehler. Alle Zellen laufen, du musst die Ergebnisse also mit Aufgabe 1 und 2 vergleichen.

```python
import numpy as np, pandas as pd, matplotlib, matplotlib.pyplot as plt, statsmodels, sklearn
import statsmodels.api as sm, statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

print('numpy', np.__version__, '| pandas', pd.__version__, '| statsmodels', statsmodels.__version__, '| scikit-learn', sklearn.__version__)
SEED = 42
np.random.seed(SEED)
Boston = pd.read_csv("Boston.csv")      # 506 tracts, 13 columns (from the ISLP package)
print(Boston.shape)
```

### Ergebnis 1: einfache Regression `medv ~ lstat`

```python
nofit = sm.OLS(Boston["medv"], Boston[["lstat"]]).fit()
print(nofit.params.round(3).to_dict(), round(nofit.rsquared, 3))
```

### Ergebnis 2: Standardfehler der Residuen

```python
fit = smf.ols("medv ~ lstat", data=Boston).fit()
n = len(Boston)
rse = np.sqrt((fit.resid ** 2).sum() / n)
print(round(rse, 2))
```

### Ergebnis 3: 95-%-Prognoseintervall bei `lstat = 10`

```python
sf = fit.get_prediction(pd.DataFrame({"lstat": [10]})).summary_frame()
print(sf[["mean_ci_lower", "mean_ci_upper"]].round(2).to_string())
```
